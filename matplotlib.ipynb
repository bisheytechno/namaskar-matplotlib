```python
import matplotlib.pyplot as plt
import numpy as np

# Model training 
epochs = [1, 2, 3, 4, 5]
loss   = [0.9, 0.7, 0.5, 0.3, 0.1]

plt.plot(epochs, loss)
plt.title("Training Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()
```

```python
#multiple lines
epochs     = [1, 2, 3, 4, 5]
train_loss = [0.9, 0.7, 0.5, 0.3, 0.1]
val_loss   = [0.95, 0.75, 0.55, 0.4, 0.3]

plt.plot(epochs, train_loss, label="Train Loss")
plt.plot(epochs, val_loss,   label="Val Loss")
plt.title("Training vs Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

```python
#bar chart

categories = ["Electronics", "Fashion", "Books", "Home"]
revenue    = [715960, 89048, 13900, 155850]

plt.bar(categories, revenue, color="steelblue")
plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue (Rs)")
plt.show()
```

```python
#histogram

marks = np.random.randint(50, 100, 100)

plt.hist(marks, bins=10, color="green")
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Count")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Plot 1
axes[0, 0].plot([1,2,3,4,5], [0.9,0.7,0.5,0.3,0.1])
axes[0, 0].set_title("Training Loss")

# Plot 2
axes[0, 1].bar(["A", "B", "C"], [85, 92, 78])
axes[0, 1].set_title("Grades")

# Plot 3
axes[1, 0].hist(np.random.randn(100), bins=15)
axes[1, 0].set_title("Distribution")

# Plot 4
axes[1, 1].scatter(
    np.random.rand(50),
    np.random.rand(50)
)
axes[1, 1].set_title("Scatter")

plt.tight_layout()
plt.show()
```
